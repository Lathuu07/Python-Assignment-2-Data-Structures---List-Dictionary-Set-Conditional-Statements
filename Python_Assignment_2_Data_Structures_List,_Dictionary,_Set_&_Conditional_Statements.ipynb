**Sets (Operations):**

a. Create a set called my_set with following values:
['a','e','i','o','u','a','a','i']


```python
my_set = {"a","e","i","o","u","a","a","i"}
```

```python
print(my_set)
```

```python
# Analysis and Explanation
# 1.created a set named my_set containing the vowels a,e,i,o,u.
# 2.The values 'a' and 'i' are repeated
# 3.In Python, a set does not allow duplicate elements.
# 4.Therefore, Python automatically removes the duplicate values.
# 5.The final output contains only unique vowels: a, e, i, o, u.
```

b. Attempt to change the value of my_set[4] = 's'. If code throws an error, provide
an explanation.

```python
my_set[4] = "s"
```

```python
# we are trying to change the value at index 4 to 's'
# sets are unordered
# indexing does not happen with sets [0],[2],[3]
# we cannot access or modify set elements using an index[4]
# so why python throws an error

#we can add or remove elements, but we cannot modify individual elements using indexing.
```

```python
print(set1)
print(set2)
```

```python
print("intersection:", set1 & set2)
```

**Performance Category Program:**

**1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).**

```python
score = int(input("Enter your score(0-10):"))
```

2. Find the performance category based on the input score using following criteria:
a. Above Average: Score greater than 7
b. Average: Score between 4 and 7(both inclusive)
c. Below Average: Score lesser than 4

```python
score = int(input("Enter your score(0-10):"))
if score > 7:
  print("Above Average")
elif score >= 4 and score <=7:
  print("Average")
else:
  print("Below Average")
```

**3. Output: Print the Performance category**

```python
score = int(input("Enter your score(0-10):"))
if score >7:
  category = "Above Average"
elif score >= 4 and score <=7:
  category = "Average"
else:
  category = "Below Average"
print("Performance Category:", category)
```

4. Additional Step: You can give a prompt of your choice to each category.
For eg: If score below average “Need to Improve your performance, consistent
practice will lead to better results”.

```python
score = int(input("Enter your score(0-10):"))
if score >7:
  category = "Above Average"
  message = "Excellent!, Keep it up."

elif score >= 4 and score <=7:
  category = "Average"
  message = "Good Work!, Keep Practicing."

else:
  category = "Below Average"
  message = "Need improvement!, Next time aim for better marks."

print("Performance Category:", category)
print("message:", message)
```
